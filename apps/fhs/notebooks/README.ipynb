## Key concepts

> **Glossary:** In [GLOSSARY.ipynb](GLOSSARY.ipynb) you will find complete definitions.

| Term | Plain English | Example |
|---|---|---|
| **Portfolio** | Your sprint or release plan viewed as a collection of features with risk and correlation | A sprint with 3 features = a portfolio of 3 investments |
| **Monte Carlo Simulation** | Run 10,000 possible futures, measure the spread of outcomes | Each scenario randomly samples users and conversions based on your uncertainty |
| **Expected Business Value** | The average outcome across all 10,000 scenarios | The mean of all simulated business values is your planning number |
| **Business Value Floor 95** | The business value floor exceeded in 95 out of 100 scenarios | "We will earn at least EUR 80,000 in 95% of cases" |
| **CVaR 95%** | Average of the worst 5% of outcomes — more conservative than the floor alone | "If things go badly, we earn EUR 65,000 on average" |
| **Risk Ratio** | Standard deviation / Expected business value — measures predictability | 15% = low risk (predictable), 40% = high risk (consider MVP first) |
| **HHI (Concentration)** | Sum of squared business value shares — measures diversification | Below 0.25 = diversified, above 0.40 = too concentrated on one bet |
| **Diversification** | Combining uncorrelated features reduces total portfolio downside | Two independent features reduce overall risk compared to one |
| **Efficient Frontier** | Portfolios with the best expected return for a given risk level | Trade-off curve: higher return requires accepting higher volatility |
