# Glossary

This glossary is a short summary of the notebook collection.
It is written for Product Owners, Portfolio Owners, and Financial Risk Managers.

The notebooks show how to evaluate features by business value,
delivery risk, downside risk, and portfolio analysis.

## How to Use This Glossary

Use the terms below as short working definitions.
They support product decisions, financial risk review,
and communication across teams.

Definitions are listed alphabetically. Each entry uses simple language first;
methodology details follow only where they help interpret the numbers.

## Break-Even Probability

Break-even probability is the share of scenarios with positive profit.
It shows how often a feature or portfolio creates value after cost.

$$P_{\text{break-even}} = \frac{\text{scenarios where profit} > 0}{\text{total scenarios}}$$

## Budget Optimizer

The budget optimizer selects the best combination of features within a fixed
investment budget. It maximises expected business value (or another chosen
strategy, e.g. business value floor) while respecting budget and risk limits.

## Business Value

Business value is the expected benefit from a feature.
It is the main value measure used in these notebooks.

## Business Value Floor (VaR 95%)

The business value floor at 95% confidence is the value exceeded in 95 out of
100 simulated scenarios. Only 5% of outcomes fall below this floor.

$$\text{Floor}_{95\%} = \text{5th percentile of all simulated outcomes}$$

If the floor is EUR 80,000, then 95% of scenarios deliver at least EUR 80,000.
Use the floor to set a minimum acceptable threshold in portfolio decisions.

## Business Value per Conversion

Business value per conversion is the expected value created each time a user
completes a key action. It equals expected users multiplied by conversion rate
and the average value per converted user.

## Conversion Rate

The conversion rate is the share of users who perform the desired action
(purchase, signup, task completion). It is one of the three core inputs when
defining a feature in FHS. A conversion rate of 0.12 means 12 out of 100 users
convert.

## Correlation

Correlation shows how strongly two variables move together.
High correlation means they often rise or fall at the same time.

## Cost at Risk (CaR 95%)

CaR 95% is the delivery cost threshold exceeded in only 5% of scenarios.
It applies the same percentile concept to cost that the business value floor
applies to value.

$$\text{CaR}_{95\%} = \text{95th percentile of simulated delivery costs}$$

## CVaR (Conditional Value at Risk)

CVaR is the average outcome in the worst part of the distribution.
It shows how severe bad outcomes can be once you cross the business value floor.

$$\text{CVaR}_{95\%} = \text{average of the worst } 5\% \text{ of simulated outcomes}$$

CVaR is always worse than or equal to the floor. Use it to see how bad things
get in the tail, not just where the tail begins.

## Delivery Risk / Bernoulli Gate (Level 2)

Delivery risk is the chance that a feature is late, costs more than planned,
or never reaches production.

In the simulations, delivery risk is applied as a per-feature gate: each feature
either delivers (with probability *p*) or is cancelled (with probability *1 − p*).
Cancelled features contribute zero business value for that scenario.

## Discount Rate

The discount rate converts future cash flows into present value.
It reflects time value and risk.

$$\text{PV} = \frac{CF}{(1+r)^t}$$

A higher discount rate means future money is worth less today.

## Diversification

Diversification means combining different features so that overall portfolio
risk is lower than the sum of individual feature risks.

## Efficient Frontier

The efficient frontier shows the best expected result for each level of risk.
Portfolios on the frontier cannot improve return without taking on more risk.

## Exact Solver

The exact solver checks every possible feature combination and returns the
guaranteed best one. Best for small portfolios (typically fewer than about
20 features).

## Expected Loss

Expected loss is the average loss across loss-making scenarios.
It tells you the typical size of a bad outcome, not how often it happens.

## Feature

A feature is a product change or initiative that may create business value
and may also add risk.

## HHI (Herfindahl-Hirschman Index)

HHI measures portfolio concentration. It is the sum of squared business value
shares across all features.

- HHI below 0.25 = well-diversified portfolio
- HHI above 0.40 = too concentrated on a single feature

High concentration means the portfolio depends heavily on one feature.
If that feature fails, the overall result suffers.

## Hypothesis

A hypothesis is a testable assumption about value, cost, or risk.
Notebooks 02–07 model each backlog candidate as a hypothesis (H1, H2, H3).

## ILP Solver

The ILP (Integer Linear Programming) solver finds near-optimal portfolios
quickly, even for larger feature sets. Use it when the exact solver becomes
too slow.

## IRR (Internal Rate of Return)

IRR is the discount rate at which net present value becomes zero.
Use it to compare investment options.

$$\text{NPV}(r = \text{IRR}) = 0$$

If IRR is above the hurdle rate, the feature creates value.

## LaR 95% (Loss at Risk)

LaR 95% is the loss threshold exceeded in only 5% of scenarios.

$$\text{LaR}_{95\%} = \text{95th percentile of the loss distribution}$$

It answers: in the worst 5% of cases, how much do we lose?

## LLP (Likelihood of Non-Delivery)

LLP is the probability that a feature fails to reach production.
At LLP 20%, one in five simulated scenarios produces zero business value
because the feature is never delivered.

LLP captures delivery risk separately from market risk. Even a high-value
feature can be a poor investment if its delivery probability is low.

## Lognormal Distribution

A lognormal distribution models positive values with a long right tail.
It is useful when rare high outcomes are possible (e.g. sprint delays,
large deal sizes).

## Monte Carlo Simulation

Monte Carlo simulation runs many random scenarios with controlled assumptions.
It produces a range of possible outcomes rather than a single point estimate.

## Normal Distribution

A normal distribution is symmetric around a mean value.
Outcomes far from the mean become less likely the further they are.

## NPV (Net Present Value)

NPV is the present value of future cash flows minus the initial investment.

$$\text{NPV} = \sum_{t=1}^{n} \frac{CF_t}{(1+r)^t} - I_0$$

where $CF_t$ is the cash flow in year $t$, $r$ is the discount rate, and
$I_0$ is the investment.

A positive NPV means the feature earns more than it costs, after adjusting
for time and risk.

## Percentile

A percentile shows the value below which a given share of results falls.
The 5th percentile of business value is the threshold exceeded in 95% of
scenarios.

## Portfolio

A portfolio is a group of features analysed together.
Portfolio analysis captures interactions between features
(e.g. shared components, correlated demand).

## Probability Distribution

A probability distribution describes how likely different outcomes are.
The notebooks use Normal, Lognormal, and Bernoulli distributions for
different inputs.

## Profitability Index (PI)

PI compares the value created to the present value of the investment:

$$\text{PI} = \frac{\text{NPV}}{\text{PV(investment)}}$$

- PI > 0 means the feature creates value beyond its investment cost.
- PI < 0 means the investment does not pay for itself.

Higher PI means more value per euro invested. PI is useful for ranking
features when the budget is limited.

## Risk

Risk is the chance of an outcome worse than expected.
The notebooks quantify risk through the business value floor, CVaR,
LaR 95%, and stress scenarios.

## Risk Ratio

The risk ratio is standard deviation divided by expected business value.
It measures how predictable a feature's outcome is.

- Risk ratio of 15% = low risk, predictable outcome
- Risk ratio of 40% = high risk, consider a prototype or MVP first

A lower risk ratio means the actual result will be closer to the expected value.

## ROI (Return on Investment)

ROI is (business value minus cost) divided by cost.
It measures how much value a feature creates relative to what it costs to build.

A positive ROI means the feature pays for itself.
A negative ROI means the feature costs more than it earns —
which may still be acceptable for regulatory or strategic features.

## Scenario

A scenario is one possible future outcome with a specific set of assumptions.
A Monte Carlo simulation generates thousands of scenarios to map the full
range of outcomes.

## Sensitivity Analysis

Sensitivity analysis shows how much a result changes when an input changes.
Use it to find the inputs that matter most.

## Shapley Value

The Shapley value fairly splits a portfolio result (value or risk) among the
contributing features. It comes from cooperative game theory and answers:
"How much did each feature contribute to the total?"

## Sharpe Ratio

The Sharpe ratio measures risk-adjusted return:
expected business value divided by its standard deviation.
A higher Sharpe ratio means more value delivered per unit of uncertainty.
Use it to compare features or portfolios with different risk profiles.

## Stress Test

A stress test checks what happens under severe but plausible negative
conditions — for example, halved conversion rates, doubled delivery delays,
or a portfolio-wide market shock.

## Sunk Cost at Cancellation

Sunk cost at cancellation is the expected spend that cannot be recovered if
the work is stopped before completion.

## Systematic Shock (Level 3)

A systematic shock is a portfolio-wide event (e.g. regulation change, recession,
supply-chain disruption) that lowers business value across many features at once.
Systematic shocks cannot be removed by diversification.

## Uncertainty

Uncertainty describes how far off an estimate could be. In FHS, it is expressed
as a fraction of the expected value.

- Uncertainty of 0.20 means the true outcome could be 20% higher or lower than estimated.
- Higher uncertainty leads to a wider range of simulated outcomes.
- Low uncertainty (< 0.15) means high confidence in the estimate.
- High uncertainty (> 0.35) means significant unknowns remain.
